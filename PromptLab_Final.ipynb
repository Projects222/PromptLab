```python
# PromptLab — AI Prompt Evaluation and Improvement System

## Project Overview

PromptLab is a Python-based prompt evaluation system designed to analyze
the quality of user prompts and identify missing components.

The system evaluates five key prompt criteria:

1. Role
2. Audience
3. Context
4. Goal
5. Output

Each criterion contributes 20 points, producing a score out of 100.

PromptLab can also:

- Identify missing prompt components
- Provide improvement suggestions
- Collect missing information from the user
- Generate an improved prompt
- Re-evaluate the improved prompt
- Calculate the improvement score
- Store prompt evaluation history
- Generate basic analytics
- Validate the system using automated tests

## Objective

The goal of PromptLab is to help users transform incomplete prompts
into clearer and more structured prompts that are more likely to produce
useful AI responses.
```

| Test | Score | Quality |
|---|---:|---|
| Weak Prompt | 20/100 | Weak |

| Complete Prompt | 100/100 | Excellent |

### Validation Result

**3/3 automated tests passed.**

The system correctly identified the presence or absence of all five
evaluation criteria.

## Limitations

PromptLab currently uses rule-based keyword detection to evaluate prompts.

For example, the system looks for phrases such as:

- "Act as" or "You are" for Role
- "Audience" for Audience
- Product-related terms for Context
- "Goal" for Goal
- "LinkedIn post" for Output

Therefore, a prompt may contain the correct information using different
wording and still be incorrectly classified as missing a criterion.

Future versions could use an AI model or semantic analysis to understand
the meaning of prompts rather than relying primarily on keyword matching.

```python
import pandas as pd

# Store PromptLab evaluation history
prompt_history = []

print("PromptLab initialized successfully! ✅")
```

```python
def get_quality_label(score):

    if score >= 90:
        return "Excellent"

    elif score >= 70:
        return "Good"

    elif score >= 50:
        return "Needs Improvement"

    else:
        return "Weak"
```

```python
def analyze_prompt(prompt):

    prompt_lower = prompt.lower()

    role = (
        "act as" in prompt_lower
        or "you are" in prompt_lower
    )

    audience = "audience" in prompt_lower

    context = (
        "product is" in prompt_lower
        or "product details" in prompt_lower
        or "product context" in prompt_lower
        or "features" in prompt_lower
        or "benefits" in prompt_lower
    )

    goal = "goal" in prompt_lower

    output = "linkedin post" in prompt_lower

    results = {
        "Role": role,
        "Audience": audience,
        "Context": context,
        "Goal": goal,
        "Output": output
    }

    score = sum(results.values()) * 20

    return results, score
    
```

```python
def display_report(results, score):

    suggestions = {
        "Role": "Specify who the AI should act as.",
        "Audience": "Define the target audience.",
        "Context": "Provide relevant background or product information.",
        "Goal": "Explain what the response should achieve.",
        "Output": "Specify the desired format, length, tone, or structure."
    }

    print("========================================")
    print("          PROMPTLAB REPORT")
    print("========================================")

    for criterion, result in results.items():

        if result:
            print(criterion, "✅ Present")

        else:
            print(criterion, "❌ Missing")
            print("Suggestion:", suggestions[criterion])

    print("========================================")
    print("Prompt Score:", score, "/100")
    print("Quality:", get_quality_label(score))
    print("========================================")
```

```python
def collect_missing_information(prompt, results):

    role = ""
    audience = ""
    context = ""
    goal = ""
    output = prompt

    print("\n========================================")
    print("       LET'S IMPROVE YOUR PROMPT")
    print("========================================")

    if not results["Role"]:
        role = input("What role should the AI act as? ")

    if not results["Audience"]:
        audience = input("Who is the target audience? ")

    if not results["Context"]:
        context = input(
            "What product details, features, or benefits should be included? "
        )

    if not results["Goal"]:
        goal = input("What is the goal of the content? ")

    return role, audience, context, goal, output
```

```python
def create_prompt(role, audience, context, goal, output):

    prompt = f"""
Act as {role}.

Target audience:
{audience}

Product context:
{context}

Goal:
{goal}

Output:
{output}
"""

    return prompt.strip()
```

```python
def save_result(
    original_prompt,
    improved_prompt,
    original_score,
    improved_score
):

    result = {
        "Original Prompt": original_prompt,
        "Original Score": original_score,
        "Improved Prompt": improved_prompt,
        "Improved Score": improved_score,
        "Improvement": improved_score - original_score
    }

    prompt_history.append(result)

    return result
```

```python
def run_promptlab():

    print("========================================")
    print("          WELCOME TO PROMPTLAB")
    print("========================================")

    # Get user's prompt
    original_prompt = input("\nEnter your prompt: ")

    # Analyze original prompt
    results, original_score = analyze_prompt(original_prompt)

    print("\n--- ORIGINAL PROMPT ANALYSIS ---")

    display_report(results, original_score)

    # If prompt is already excellent
    if original_score == 100:

        print("\nYour prompt already meets all five criteria! ✅")
        print("No improvement is required.")

        return original_prompt

    # Collect missing information
    role, audience, context, goal, output = (
        collect_missing_information(
            original_prompt,
            results
        )
    )

    # Build improved prompt
    improved_prompt = create_prompt(
        role,
        audience,
        context,
        goal,
        output
    )

    # Analyze improved prompt
    improved_results, improved_score = analyze_prompt(
        improved_prompt
    )

    # Save result
    save_result(
        original_prompt,
        improved_prompt,
        original_score,
        improved_score
    )

    # Final report
    print("\n========================================")
    print("          PROMPTLAB FINAL RESULT")
    print("========================================")

    print("\nOriginal Score:",
          original_score, "/100")

    print("Improved Score:",
          improved_score, "/100")

    print("Improvement:",
          improved_score - original_score,
          "points")

    print("Quality:",
          get_quality_label(improved_score))

    print("\n--- IMPROVED PROMPT ---")
    print(improved_prompt)

    print("\n✅ Result saved to PromptLab history.")

    print("========================================")

    return improved_prompt
```

```python
run_promptlab()
```

```python
def run_test_suite():

    print("========================================")
    print("        PROMPTLAB TEST SUITE")
    print("========================================")

    test_prompts = [

        # TEST 1: Weak prompt
        "Write a LinkedIn post about FocusPro X1.",

        # TEST 2: Medium prompt
        "Act as a marketing specialist. "
        "Write a LinkedIn post about FocusPro X1.",

        # TEST 3: Excellent prompt
        "Act as a social media marketing specialist. "
        "The target audience is students, YouTube bloggers, "
        "music lovers, and teachers. "
        "The product is FocusPro X1, a lightweight headphone "
        "with active noise cancellation, 40-hour battery life, "
        "and an excellent microphone for meetings. "
        "The goal is to drive traffic to the product page. "
        "Write a LinkedIn post about FocusPro X1."
    ]

    for number, prompt in enumerate(test_prompts, start=1):

        print("\n")
        print("========================================")
        print("TEST", number)
        print("========================================")

        print("Prompt:")
        print(prompt)

        results, score = analyze_prompt(prompt)

        print("\nResults:")

        for criterion, result in results.items():

            if result:
                print(criterion, "✅ Present")
            else:
                print(criterion, "❌ Missing")

        print("\nScore:", score, "/100")
        print("Quality:", get_quality_label(score))

    print("\n========================================")
    print("       TEST SUITE COMPLETE")
    print("========================================")
```

```python
run_test_suite()
```

```python
def validate_promptlab():

    test_cases = [
        (
            "Weak prompt",
            "Write a LinkedIn post about FocusPro X1.",
            20
        ),
        (
            "Partial prompt",
            "Act as a marketing specialist. "
            "Write a LinkedIn post about FocusPro X1.",
            40
        ),
        (
            "Complete prompt",
            "Act as a social media marketing specialist. "
            "The target audience is students, YouTube bloggers, "
            "music lovers, and teachers. "
            "The product is FocusPro X1, a lightweight headphone "
            "with active noise cancellation, 40-hour battery life, "
            "and an excellent microphone for meetings. "
            "The goal is to drive traffic to the product page. "
            "Write a LinkedIn post about FocusPro X1.",
            100
        )
    ]

    passed = 0

    print("========================================")
    print("       PROMPTLAB VALIDATION")
    print("========================================")

    for name, prompt, expected_score in test_cases:

        results, actual_score = analyze_prompt(prompt)

        if actual_score == expected_score:
            print(name, "✅ PASS")
            passed += 1
        else:
            print(
                name,
                "❌ FAIL",
                "| Expected:",
                expected_score,
                "| Got:",
                actual_score
            )

    print("----------------------------------------")
    print("Tests Passed:", passed, "/", len(test_cases))

    if passed == len(test_cases):
        print("STATUS: ALL TESTS PASSED ✅")
    else:
        print("STATUS: SOME TESTS FAILED ❌")

    print("========================================")
```

```python
validate_promptlab()
```
